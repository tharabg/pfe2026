```python
def is_identity_document(text):
    keywords = [
        "identity", "national", "id", "name",
        "date", "birth", "card", "passport"
    ]
    score = sum(1 for k in keywords if k.lower() in text.lower())
    return score >= 2
```

```python
def preprocess_image(path):
    img = cv2.imread(path)

    # Redimensionnement (standard)
    img = cv2.resize(img, (800, 500))

    # Gris
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # Réduction du bruit
    blur = cv2.GaussianBlur(gray, (5,5), 0)

    # Binarisation
    thresh = cv2.adaptiveThreshold(
        blur, 255,
        cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
        cv2.THRESH_BINARY,
        11, 2
    )

    return thresh
```

```python
from PIL import Image, ImageDraw, ImageFont
import os

def create_recto(cin, output_path):
    img = Image.new("RGB", (800, 500), color="white")
    d = ImageDraw.Draw(img)

    d.text((30, 30), "CARTE D'IDENTITE NATIONALE", fill="black")
    d.text((30, 100), f"NOM : {cin['nom']}", fill="black")
    d.text((30, 140), f"PRENOM : {cin['prenom']}", fill="black")
    d.text((30, 180), f"DATE NAISSANCE : {cin['date_naissance']}", fill="black")
    d.text((30, 220), f"CIN : {cin['cin_number']}", fill="black")

    img.save(output_path)
```

```python
def create_verso(cin, output_path):
    img = Image.new("RGB", (800, 500), color="white")
    d = ImageDraw.Draw(img)

    d.text((30, 50), "ADRESSE :", fill="black")
    d.text((30, 90), cin["adresse"], fill="black")
    d.text((30, 160), f"PAYS : {cin['pays']}", fill="black")
    d.text((30, 220), f"NUMERO : {cin['cin_number']}", fill="black")

    img.save(output_path)
```
